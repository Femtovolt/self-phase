```python
import os
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

# Define the path to the data folder
data_folder = "../data"  # Adjust this to your data folder path
# Placeholder for plots from all subfolders
gas_energy_groups = {}
combined_energy_groups = {}
# Traverse all subfolders in the data directory
i=0
for subfolder in os.listdir(data_folder):
    subfolder_path = os.path.join(data_folder, subfolder)
    # Read the params file
    params_path = os.path.join(subfolder_path, "params")
    with open(params_path, "r") as f:
        params = f.readlines()
    # Extract dz and Energy from the params file
    dz = None
    energy = None
    for line in params:
        if line.startswith("dz:"):
            dz = float(line.split()[1].strip("[]"))
        if line.startswith("Energy:"):
            energy = float(line.split()[1].strip("[]"))*10**6
        elif line.startswith("Gas:"):
            gas = line.split()[1].strip()

    # Read the PlasmaDensity file
    plasma_density_path = os.path.join(subfolder_path, "PlasmaDensity")
    with open(plasma_density_path, "r") as f:
        plasma_density = [float(line.strip()) for line in f.readlines()]

    # Prepare x and y values for plotting
    x_values = [i * dz for i in range(len(plasma_density))]
    y_values = plasma_density

    # Group by Gas and Energy
    if gas not in gas_energy_groups:
        gas_energy_groups[gas] = {}
    if energy not in gas_energy_groups[gas]:
        gas_energy_groups[gas][energy] = []
        gas_energy_groups[gas][energy].append((subfolder, x_values, y_values))

for gas, energy_groups in gas_energy_groups.items():
    for energy, plots in energy_groups.items():
        if energy not in combined_energy_groups:
            combined_energy_groups[energy] = {}
        combined_energy_groups[energy][gas] = plots

sorted_energies = sorted(combined_energy_groups.items(), key=lambda x: x[0], reverse=True)
# Plotting
num_plots = len(sorted_energies)
fig, axes = plt.subplots(num_plots, 1, figsize=(12, 4 * num_plots), sharex=True)
if num_plots == 1:
    axes = [axes]

for ax, (energy, gas_groups) in zip(axes, sorted_energies):
    for gas, plots in gas_groups.items():
        for subfolder, x_values, y_values in plots:
            max_density = max(y_values)
            ax.plot(x_values, y_values, label=f"{gas}(Max: {max_density:.2e})")
    ax.set_title(f"Energy: {energy} uJ")
    ax.grid()
    ax.legend()
    # Force scientific notation
    scalar_formatter = ticker.ScalarFormatter(useMathText=True)
    scalar_formatter.set_powerlimits((0,0))  
    ax.yaxis.set_major_formatter(scalar_formatter)

fig.text(0.04, 0.5, "Plasma Density (1/cm)", va='center', rotation='vertical', fontsize=20)
plt.ylabel(" ")
plt.xlabel("Position m")
plt.tight_layout(rect=[0.05, 0, 1, 1])
plt.show()
```
